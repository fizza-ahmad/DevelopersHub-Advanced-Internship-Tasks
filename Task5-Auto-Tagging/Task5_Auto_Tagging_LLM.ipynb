```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

```python
labels = [
    "Billing Issue",
    "Technical Issue",
    "Password Reset",
    "Account Access",
    "Refund Request",
    "Delivery Issue",
    "Complaint",
    "Subscription Issue"
]
```

```python
ticket = "I cannot login to my account after resetting my password"

result = classifier(ticket, labels)

print(result)
```

```python
top_3 = result["labels"][:3]

print("Top 3 Predicted Tags:")
for tag in top_3:
    print("-", tag)
```

```python
few_shot_prompt = """
Example 1:
Ticket: "I was charged twice for my subscription"
Tag: Billing Issue

Example 2:
Ticket: "My app is crashing when I open it"
Tag: Technical Issue

Example 3:
Ticket: "I forgot my password and cannot login"
Tag: Password Reset

Now classify:
Ticket: I cannot access my account after login attempt
"""
result2 = classifier(few_shot_prompt, labels)

print(result2["labels"][:3])

print("Zero-shot:", result["labels"][:3])
print("Few-shot:", result2["labels"][:3])
```
